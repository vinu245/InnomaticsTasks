## Academic evaluation systems


```python
marks=[45, 78, 90, 33, 60]
Pass=0
Fail=0
for i in marks:
    if i >= 50:
        print ("Student Pass")
        Pass += 1
    else:
        print("Student Fail")
        Fail+=1
print(f"Toatal number of students passed are {Pass}")
print(f"Toatal number of students Failed are {Fail}")
```

## Data preprocessing before analysis


```python
names = [" Alice ", "bob", " CHARLIE "]
cleaned_names=[]
for i in names:
    #remove extra spaces .strip()
    #convert to lowercases .lower()
    clean_names= i.strip().lower()
    #append is used to add / insert values into list
    cleaned_names.append(clean_names)
print(cleaned_names)
```
