# Problem 1: Smart Light Controller

```python
class SmartLight:
    
    # Constructor to initialize light name and default state
    def __init__(self, light_name):
        self.light_name = light_name
        self.status = "OFF"   # Default state
    
    # Method to turn ON the light
    def turn_on(self):
        self.status = "ON"
        print(f"{self.light_name} is ON")
    
    # Method to turn OFF the light
    def turn_off(self):
        self.status = "OFF"
        print(f"{self.light_name} is OFF")
    
    # Method to display current status
    def display_status(self):
        print(f"{self.light_name} is {self.status}")
    # Create object
light1 = SmartLight("Bedroom Light")
light2= SmartLight("Kitchen Light")

# Perform action
light1.turn_on()
light2.turn_off()
```

# Problem 2: Employee ID Card System

```python
class Employee:
    
    # Constructor to initialize employee details
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department
    
    # Method to display ID card
    def display_id_card(self):
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")
# Creating employee object
emp1 = Employee("vinu", "1de233", "Ise")

# Display ID Card
emp1.display_id_card()
```

# Problem 3: Mobile Contact Record

```python
class Contact:
    
    # Constructor to initialize contact details
    def __init__(self, name, phone):
        self.name = name
        self.phone = phone
    
    # Method to display contact information
    def display_contact(self):
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone}")
# Create contact object
contact1 = Contact("vishwas", "9876543210")
contact2 = Contact("vinutha", "9866543710")

# Display contact information
contact1.display_contact()
contact2.display_contact()
```

# Problem 4: Product Price Tag Generator

```python
class Product:
    
    # Constructor
    def __init__(self, name, price):
        self.name = name
        self.price = price
    
    # Method to print price tag
    def print_price_tag(self):
        print(f"Product: {self.name}")
        print(f"Price: ₹{self.price}")
product1 = Product("Headphones", 2499)
product1.print_price_tag()
```

# Problem 5: Movie Rating Display System

```python
class Movie:
    
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating
    
    def display_movie(self):
        print(f"Movie: {self.name}")
        print(f"Rating: {self.rating} / 5")
movie1 = Movie("Inception", 4.8)
movie1.display_movie()
```

# Problem 6: Delivery Address Manager

```python
class Delivery:
    
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address
    
    def print_delivery_details(self):
        print("Delivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")
delivery1 = Delivery("Suman", "Hyderabad")
delivery1.print_delivery_details()
```
