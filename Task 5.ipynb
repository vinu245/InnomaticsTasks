```python
def analyze_engagement(likes_list):
    total_likes = 0
    
    # Calculate total likes using loop
    for like in likes_list:
        total_likes += like
    
    # Check post status
    if total_likes >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    
    # Display result
    print(f"Total Likes: {total_likes}")
    print(f"Post Status: {status}")


# Example Usage
likes = [250, 300, 150, 200, 150]  # Total = 1050
analyze_engagement(likes)
```

```python
def check_rainfall(rainfall_data, required_level):
    
    total = 0
    
    # Calculate total rainfall using loop
    for rain in rainfall_data:
        total += rain
    
    # Calculate average
    average = total / len(rainfall_data)
    
    # Check adequacy
    if average >= required_level:
        status = "Adequate Rainfall"
    else:
        status = "Inadequate Rainfall"
    
    # Display result
    print(f"Average Rainfall: {int(average)}")
    print(f"Rainfall Status: {status}")


# Example Usage
rainfall = [60, 80, 75, 70, 75]   # Average = 72
required = 70

check_rainfall(rainfall, required)
```

```python
def detect_duplicates(usernames):
    if len(usernames) != len(set(usernames)):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")

accounts = ["user1", "user2", "user3", "user1"]
detect_duplicates(accounts)
```

```python
def filter_premium_crops(prices):
    
    premium_crops = []
    
    for price in prices:
        if price > 2000:
            premium_crops.append(price)
    
    print(f"Premium Crops: {premium_crops}")


# Example
crop_prices = [1500, 2500, 1800, 3200]
filter_premium_crops(crop_prices)
```

```python
def check_exam_eligibility(attendance_list):
    
    total_classes = len(attendance_list)
    attended = 0
    
    for status in attendance_list:
        if status == 1:   # 1 = Present, 0 = Absent
            attended += 1
    
    percentage = (attended / total_classes) * 100
    
    if percentage >= 75:
        eligibility = "Eligible"
    else:
        eligibility = "Not Eligible"
    
    print(f"Attendance Percentage: {percentage}")
    print(f"Exam Eligibility: {eligibility}")


# Example
attendance = [1,1,1,0,1,1,1,1,0,1]  # 80%
check_exam_eligibility(attendance)
```

```python
def calculate_electricity_bill(units):
    
    bill = 0
    
    # Slab calculation
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)
    
    # Usage classification
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"
    
    return bill, usage


# Example
total_bill, status = calculate_electricity_bill(250)

print(f"Total Bill: ₹{total_bill}")
print(f"Usage Level: {status}")
```
