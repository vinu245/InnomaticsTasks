# Problem 1: Smart Parking Lot Management System

```python
def smart_parking(capacity, logs):
    current = 0
    peak = 0

    for log in logs:
        if log == "IN":
            current += 1
        elif log == "OUT" and current > 0:
            current -= 1

        if current > peak:
            peak = current

    print("Currently Parked Vehicles:", current)

    if current > capacity:
        print("Parking Status: Full - Alert!")
    else:
        print("Parking Status: Available")

    print("Peak Parking Usage:", peak)
smart_parking(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

# Problem 2: Online Food Delivery Time Estimator

```python
def estimate_delivery(distance, traffic, weather):
    time = distance * 5   # base time

    # Traffic adjustment
    if traffic == "High":
        time += 15
    elif traffic == "Medium":
        time += 8

    # Weather adjustment
    if weather == "Rainy":
        time += 10
    elif weather == "Storm":
        time += 20

    print("Estimated Delivery Time:", time, "minutes")
estimate_delivery(8, "High", "Rainy")
```

## Problem 3: Movie Theatre Seat Occupancy Analyzer

```python
def theatre_occupancy(total_seats, booked_seats):
    booked = len(booked_seats)
    percentage = (booked / total_seats) * 100

    print("Occupancy:", int(percentage), "%")

    if percentage == 100:
        print("Show Status: Housefull")
    elif percentage >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Available")
theatre_occupancy(200, [1]*150)
```

## Problem 4: Cloud Server Load Classification System

```python
def server_load(cpu_readings):
    avg = sum(cpu_readings) / len(cpu_readings)

    print("Average CPU Load:", int(avg), "%")

    if avg < 50:
        print("Server Status: Normal")
    elif avg <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")
server_load([45, 60, 70, 85, 90])
```

## Problem 5: Smart Classroom Resource Usage Monitor

```python
def classroom_monitor(resources):
    overused = []

    for resource, hours in resources.items():
        if hours > 8:
            overused.append(resource)

    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")
classroom_monitor({
    "Projector": 6,
    "AC": 9,
    "Lights": 4
})
```

## Problem 6: Online Event Registration Capacity Controller

```python
def event_registration(capacity, registrations):
    if registrations > capacity:
        confirmed = capacity
        waitlist = registrations - capacity
        status = "Closed"
    else:
        confirmed = registrations
        waitlist = 0
        status = "Open"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlist)
    print("Registration Status:", status)
event_registration(100, 105)
```
